```python
import pandas as pd
import numpy as np
```

Задание 7.1.1

```python
universities = pd.Series(
    data=['МГУ', 'СПбГУ', 'МФТИ', 'ВШЭ', 'МГТУ им. Баумана'],
    index=[101, 102, 103, 104, 105],
    dtype=str,
    name='Университеты'
)
print(universities)
```

Задание 7.1.2

```python
cities = pd.Series(
    data=['Москва', 'Санкт-Петербург', 'Долгопрудный', 'Москва', 'Москва'],
    index=[101, 102, 103, 104, 105],
    dtype=str,
    name='Города'
)
print(cities)
```

Задание 7.1.3

```python
students = pd.Series(
    data=[40000.0, 32000.0, 6500.0, 50000.0, 19000.0],
    index=[101, 102, 103, 104, 105],
    dtype=float,
    name='Количество студентов'
)
print(students)
```

Задание 7.1.4

```python
uni_city = pd.Series(
    data=universities.astype(str) + ' — ' + cities.astype(str),
    index=universities.index,
    dtype=str,
    name='Университет и город'
)
print(uni_city)
```

Задание 7.1.5

```python
max_students = students.max()
min_students = students.min()
mean_students = students.mean()

print(f'Максимальное количество студентов: {max_students}')
print(f'Минимальное количество студентов:  {min_students}')
print(f'Среднее количество студентов:     {mean_students}')

print('\nТипы результатов:')
print('max:', type(max_students), '| dtype:', np.dtype(type(max_students)))
print('min:', type(min_students))
print('mean:', type(mean_students))
```

Задание 7.2.1

```python
df1 = pd.DataFrame(
    data={
        'Год основания': [1755, 1724, 1946, 1992, 1830],
        'Рейтинг':       [1, 2, 3, 4, 5],
        'Факультетов':   [40, 25, 12, 18, 20],
    },
    index=[101, 102, 103, 104, 105]
)
print(df1)
```

Задание 7.2.2

```python
df2 = pd.DataFrame({
    universities.name: universities,
    cities.name:       cities,
    students.name:     students,
})
print(df2)
```

Задание 7.2.3

```python
df3 = pd.concat([df1, df2], axis=1)
print(df3)
```

Задание 7.2.4

```python
print('Метки строк (index):')
print(df3.index.tolist())
print('\nМетки столбцов (columns):')
print(df3.columns.tolist())
print('\nМассив данных (values):')
print(df3.values)
```

Задание 7.2.5

```python
# А
print(df3.head(2))
# Б
print(df3.iloc[1:4])
# В
print(df3[universities.name])
# Г
print(df3[students.name])
# Д
target_id = 103
print(f'ID = {target_id} → {df3.loc[target_id, universities.name]}')
# Е
target_pos = 3
print(f'Позиция {target_pos} → {df3.iloc[target_pos][cities.name]}')
```
