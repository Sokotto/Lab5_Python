Задание 7.3.1

```python
text = '''"Дата","Цена","Откр.","Макс.","Мин.","Объём","Изм. %"
"20.10.2026","251,30","250,80","252,10","249,90","2,35M","0,45%"
"19.10.2026","250,17","249,50","251,00","248,70","3,10M","-0,32%"
"16.10.2026","-","251,00","252,40","249,80","1,98M","0,12%"
"15.10.2026","250,50","249,90","251,70","249,10","2,76M","0,80%"
"14.10.2026","248,48","247,60","249,30","246,90","4,05M","-0,55%"
"13.10.2026","249,85"," ","н/д","247,20","3,42M","0,27%"
"12.10.2026","249,17","248,00","250,10","247,50","2,11M","0,91%"
"09.10.2026","246,92","246,10","247,80","245,40","1,87M","-0,18%"
"08.10.2026","247,37","246,80","248,20","245,90","-","0,36%"
"07.10.2026","246,48","245,70","247,40","244,80","2,64M","0,05%"
'''
with open("stock.csv", "w", encoding="utf-8") as f:
    f.write(text)

def convert_date(s):
    if s.strip() in ["", "-", "н/д"]:
        return pd.NaT
    return pd.to_datetime(s.strip(), format="%d.%m.%Y")

def convert_volume(s):
    s = s.strip()
    if s in ["", "-", "н/д"]:
        return float("nan")
    if s.endswith("M"):
        return float(s[:-1].replace(",", ".")) * 1000000
    if s.endswith("K"):
        return float(s[:-1].replace(",", ".")) * 1000
    return float(s.replace(",", "."))

df = pd.read_csv(
    "stock.csv",
    header=0,
    names=["Date", "Close", "Open", "High", "Low", "Volume"],
    usecols=[0, 1, 2, 3, 4, 5],
    dtype={"Close": "float64", "Open": "float64",
           "High": "float64", "Low": "float64"},
    converters={"Date": convert_date,
                "Volume": convert_volume},
    na_values=["-", " ", "н/д"],
    decimal=",",
    encoding="utf-8"
)

df["Date"] = df["Date"].astype("datetime64[ns]")
df["Volume"] = df["Volume"].astype("float64")

print(df)
print()
print(df.dtypes)
print()
print("Пропуски по столбцам:")
print(df.isna().sum())
```

Задание 7.3.2

```python
# а) Сортировка по объёму продаж по убыванию
df_sorted = df.sort_values("Volume", ascending=False)
print("а) Сортировка по Volume (по убыванию):")
print(df_sorted)
print()

# б) Ранжирование по цене открытия
df_rank = df.copy()
df_rank["Open_rank"] = df_rank["Open"].rank()   # 1 - самая низкая цена открытия
print("б) Ранги цены открытия:")
print(df_rank[["Date", "Open", "Open_rank"]])
print()

# в) Сводная статистика по ценам
price_cols = ["Close", "Open", "High", "Low"]
print("в) describe() по ценам:")
print(df[price_cols].describe())
print()

# г) Корреляция и ковариация: цена закрытия и объём
print("г) Корреляция Close и Volume:", df["Close"].corr(df["Volume"]))
print("   Ковариация Close и Volume:", df["Close"].cov(df["Volume"]))
print()
print("Корреляционная матрица:")
print(df[["Close", "Volume"]].corr())
print()
print("Ковариационная матрица:")
print(df[["Close", "Volume"]].cov())
print()

# д) Обработка пропусков
print("д) Пропусков в каждой строке до обработки:")
print(df.isna().sum(axis=1))
print()

df_drop = df.dropna(thresh=5)
print("Размер до удаления:", df.shape, ", после:", df_drop.shape)
print(df_drop)
print()

df_fill = df_drop.copy()

df_fill[price_cols] = df_fill[price_cols].apply(
    lambda row: row.fillna(row.mean()), axis=1
)
df_fill["Volume"] = df_fill["Volume"].fillna(df_fill["Volume"].mean())

print("Данные после заполнения пропусков:")
print(df_fill)
print()
print("Пропусков осталось:")
print(df_fill.isna().sum())
```

```python
text = '''"Дата","Цена","Откр.","Макс.","Мин.","Объём","Изм. %"
"20.10.2025","231,40","230,50","232,30","229,80","2,05M","0,52%"
"17.10.2025","230,20","229,70","231,10","228,90","2,88M","-0,21%"
"16.10.2025","230,68","229,40","231,50","228,60","1,76M","0,33%"
"15.10.2025","229,92","228,80","230,70","228,10","3,12M","0,41%"
"14.10.2025","228,98","228,00","229,90","227,30","2,47M","-0,15%"
"13.10.2025","229,33","228,40","230,20","227,90","2,93M","0,22%"
"10.10.2025","228,83","227,60","229,70","226,80","-","0,60%"
"09.10.2025","227,46","226,90","228,40","226,10","1,64M","-0,29%"
"08.10.2025","228,12","227,20","229,00","226,50","2,21M","0,18%"
"07.10.2025","227,71","226,70","228,60","226,00","2,56M","0,07%"
'''
with open("stock_2025.csv", "w", encoding="utf-8") as f:
    f.write(text)
df_old = pd.read_csv(
    "stock_2025.csv",
    header=0,
    names=["Date", "Close", "Open", "High", "Low", "Volume"],
    usecols=[0, 1, 2, 3, 4, 5],
    dtype={"Close": "float64", "Open": "float64",
           "High": "float64", "Low": "float64"},
    converters={"Date": convert_date,
                "Volume": convert_volume},
    na_values=["-", " ", "н/д"],
    decimal=",",
    encoding="utf-8"
)
df_old["Date"] = df_old["Date"].astype("datetime64[ns]")
df_old["Volume"] = df_old["Volume"].astype("float64")

print("Данные за 2026 год:", df.shape)
print("Данные за 2025 год:", df_old.shape)
print()

# а) Конкатенация: записи одного набора под записями другого
df_concat = pd.concat([df, df_old], ignore_index=True)   # индекс делаем заново
df_concat = df_concat.sort_values("Date", ascending=False).reset_index(drop=True)

print("а) concat():")
print(df_concat)
print("Размер:", df_concat.shape)
print()

df_concat_keys = pd.concat([df, df_old], keys=["2026", "2025"])
print("а) concat() с keys:")
print(df_concat_keys.head(12))
print()

# б) merge(): соединение по общему ключу
df["DayMonth"] = df["Date"].dt.strftime("%d.%m")
df_old["DayMonth"] = df_old["Date"].dt.strftime("%d.%m")

df_inner = pd.merge(df, df_old, on="DayMonth", how="inner",
                    suffixes=("_2026", "_2025"))
print("б) merge(), how='inner':")
print(df_inner[["DayMonth", "Close_2026", "Close_2025", "Volume_2026", "Volume_2025"]])
print("Размер:", df_inner.shape)
print()

df_outer = pd.merge(df, df_old, on="DayMonth", how="outer",
                    suffixes=("_2026", "_2025"))
print("б) merge(), how='outer':")
print(df_outer[["DayMonth", "Close_2026", "Close_2025"]])
print("Размер:", df_outer.shape)
print()

df_inner["Close_diff"] = df_inner["Close_2026"] - df_inner["Close_2025"]
print("Разница цены закрытия (2026 - 2025):")
print(df_inner[["DayMonth", "Close_diff"]])
```

```python
text = '''"Дата","Цена","Откр.","Макс.","Мин.","Объём","Изм. %"
"15.10.2026","250,50","249,90","251,70","249,10","2,76M","0,80%"
"14.10.2026","248,48","247,60","249,30","246,90","4,05M","-0,55%"
"13.10.2026","249,85"," ","н/д","247,20","3,42M","0,27%"
"12.10.2026","249,17","248,00","250,10","247,50","2,11M","0,91%"
"09.10.2026","246,92","246,10","247,80","245,40","1,87M","-0,18%"
"08.10.2026","247,37","246,80","248,20","245,90","-","0,36%"
"07.10.2026","246,48","245,70","247,40","244,80","2,64M","0,05%"
"06.10.2026","246,05","245,40","246,90","244,90","1,95M","0,10%"
"05.10.2026","245,95","245,20","246,70","244,60","2,38M","0,24%"
"02.10.2026","245,10","244,30","246,00","243,50","2,02M","-0,12%"
'''
with open("stock_2026_new.csv", "w", encoding="utf-8") as f:
    f.write(text)
df_new = pd.read_csv(
    "stock_2026_new.csv",
    header=0,
    names=["Date", "Close", "Open", "High", "Low", "Volume"],
    usecols=[0, 1, 2, 3, 4, 5],
    dtype={"Close": "float64", "Open": "float64",
           "High": "float64", "Low": "float64"},
    converters={"Date": convert_date,
                "Volume": convert_volume},
    na_values=["-", " ", "н/д"],
    decimal=",",
    encoding="utf-8"
)
df_new["Date"] = df_new["Date"].astype("datetime64[ns]")
df_new["Volume"] = df_new["Volume"].astype("float64")

cols = ["Date", "Close", "Open", "High", "Low", "Volume"]
df_old = df[cols].copy()

print("Исходный набор:", df_old.shape)
print("Новый набор:", df_new.shape)
print("Общих дат:", df_old["Date"].isin(df_new["Date"]).sum())
print()

# а) Конкатенация без дублей
df_concat = pd.concat([df_old, df_new], ignore_index=True)
print("а) concat() без удаления дублей:", df_concat.shape)
print("   Повторяющихся дат:", df_concat["Date"].duplicated().sum())

df_concat = df_concat.drop_duplicates(subset="Date", keep="first")
df_concat = df_concat.sort_values("Date", ascending=False).reset_index(drop=True)

print("   После drop_duplicates():", df_concat.shape)
print(df_concat)
print()

# б) merge() без дублей
df_merge = pd.merge(df_old, df_new, how="outer")
df_merge = df_merge.sort_values("Date", ascending=False).reset_index(drop=True)

print("б) merge(how='outer'):", df_merge.shape)
print("   Повторяющихся дат:", df_merge["Date"].duplicated().sum())
print(df_merge)
print()

print("Наборы равны:", df_concat.equals(df_merge))
```
